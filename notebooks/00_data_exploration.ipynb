```python
important_columns = [
    "Hugo_Symbol",
    "NCBI_Build",
    "Chromosome",
    "Start_Position",
    "End_Position",
    "Reference_Allele",
    "Tumor_Seq_Allele2",
    "Variant_Classification",
    "Variant_Type",
    "Tumor_Sample_Barcode",
    "HGVSc",
    "HGVSp_Short",
    "Transcript_ID",
    "GDC_FILTER"
]

maf[important_columns].head()


# The theory behind each selected column:
    #Hugo_Symbol: gene name, such as TP53
    #NCBI_Build: genome assembly
    #Chromosome: chromosome
    #Start_Position, End_Position: genomic coordinates
    #Reference_Allele: original DNA base
    #Tumor_Seq_Allele2: mutated DNA base
    #Variant_Classification: consequence, such as missense
    #Variant_Type: SNP, insertion, deletion
    #Tumor_Sample_Barcode: tumour sample ID
    #HGVSc: coding DNA mutation
    #HGVSp_Short: protein mutation
    #Transcript_ID: transcript affected
    #GDC_FILTER: quality-related annotation
```

```python
maf["Variant_Classification"].value_counts()
    # describes the biological consequence of each mutation
    #Missense_Mutation: changes one amino acid
    #Silent: DNA changes, but amino acid stays the same
    #Nonsense_Mutation: creates a premature stop codon
    #Frame_Shift_Del: deletion shifts the reading frame
    #Splice_Site: may affect RNA splicing
    #Intron: located in a non-coding intron
```

```python
# PASS is encoded as an empty GDC_FILTER in this MAF; the literal
# string "PASS" never appears (see the value_counts cell above).
filtered_maf = maf[
    (maf["Variant_Classification"] == "Missense_Mutation") &
    (maf["Variant_Type"] == "SNP") &
    (maf["GDC_FILTER"].isna() | (maf["GDC_FILTER"] == ""))
].copy()
```
